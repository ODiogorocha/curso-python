## Estruturas de Dados em Python
Python possui estruturas como listas, tuplas, conjuntos e dicionários.

```python
# Cria uma lista (mutável) com três elementos inteiros
lista = [10, 20, 30]
# Adiciona o valor 40 ao final da lista usando o método append
lista.append(40)
# Imprime a lista completa
print('Lista:', lista)

# Define uma tupla (imutável) com três strings
tupla = ('maçã', 'banana', 'laranja')
# Imprime a tupla. Tuplas não podem ser alteradas depois de criadas
print('Tupla:', tupla)

# Cria um conjunto (set). Valores duplicados são removidos automaticamente
conjunto = {1, 2, 2, 3}
# Imprime o conjunto — a ordem não é garantida e o 2 aparecerá apenas uma vez
print('Conjunto:', conjunto)

# Cria um dicionário com chaves e valores (estrutura de mapeamento)
dicionario = {'nome': 'Carlos', 'idade': 25}
# Imprime o dicionário inteiro
print('Dicionário:', dicionario)
# Acessa um valor pelo nome da chave (aqui acessamos a idade)
print('Idade:', dicionario['idade'])
```

# Módulo 3 – Programação Orientada a Objetos em Python
Nesta aula, vamos estudar os **conceitos fundamentais da Programação Orientada a Objetos (POO)** em Python.
Você aprenderá a criar classes, instanciar objetos e aplicar princípios como encapsulamento, herança, polimorfismo e abstração.


## 1. Conceitos Fundamentais
- **Classe**: molde ou estrutura que define como os objetos serão criados.
- **Objeto**: instância concreta da classe, algo que existe na memória.
- **Atributos**: características do objeto (dados armazenados).
- **Métodos**: comportamentos ou ações que o objeto pode executar.


```python
class Carro:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def exibir_info(self):
        print(f'Carro: {self.marca} {self.modelo}')

# Criando um objeto (instância da classe)
meu_carro = Carro('Toyota', 'Corolla')
meu_carro.exibir_info()
```

## 2. Classe em Python
- Definimos uma classe usando a palavra-chave `class`.
- O método especial `__init__` é o **construtor**, executado quando o objeto é criado.
- `self` representa a própria instância da classe.


## 3. Encapsulamento
O encapsulamento protege os dados de acessos indevidos.

Convenções em Python:
- `atributo`: público
- `_atributo`: protegido (uso interno)
- `__atributo`: privado (não acessível diretamente)


```python
class ContaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular        # público
        self._saldo = saldo          # protegido
        self.__senha = '1234'        # privado

    def exibir_saldo(self):
        print(f'Saldo de {self.titular}: R${self._saldo}')

conta = ContaBancaria('Ana', 1000)
conta.exibir_saldo()
# print(conta.__senha) # Isso daria erro, atributo privado
```

##4. Herança
Uma classe pode **herdar atributos e métodos** de outra.
- Permite reuso de código.
- Classes filhas podem sobrescrever métodos da classe pai.


```python
class Animal:
    def falar(self):
        print('O animal faz um som')

class Cachorro(Animal):
    def falar(self):
        print('Au au!')

class Gato(Animal):
    def falar(self):
        print('Miau!')

animais = [Cachorro(), Gato(), Animal()]
for a in animais:
    a.falar()
```

## 5. Polimorfismo
O mesmo método pode ter **diferentes comportamentos** em classes distintas.
No exemplo acima, o método `falar()` foi sobrescrito em cada classe filha.


## 6. Abstração
A abstração simplifica sistemas complexos, mostrando apenas o essencial.
Em Python, usamos o módulo `abc` para criar **classes abstratas**.


```python
from abc import ABC, abstractmethod

class Forma(ABC):
    @abstractmethod
    def area(self):
        pass

class Quadrado(Forma):
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2

q = Quadrado(4)
print('Área do quadrado:', q.area())
```

## 7. Vantagens da POO
- Organização do código: estrutura clara e modular.
- Reutilização: herança e composição evitam duplicação.
- Facilidade de manutenção: alterações localizadas.
- Modelagem próxima ao mundo real: objetos representam entidades reais.


# Desafio:
1. classe aluno

Atribnutos: nome, idade, curso, nomes(listas).
metodos adicionar_nota(nota) : Adiciona uma nota a lista
metodo media(): Retorna a media das notas
metodo exibir_info(): mostra nome, idade, curso e media

2. classe Turma
Atributos: nome_turma, alunos(lista).

metodo adicionar_aluno(aluno): Adiciona um objeto aluno a turma.

metodo listar_alunos(): mostra todos os alunos com suas medias.
metodo melhor_aluno(): Retorna o aluno com maior media

3. Fluxo do Programa
criar pelo menos 3 alunos
adicionar notas a cada aluno
inserir os alunos em uma turma
exibir os alunos da turma com suas medias
mostrar o melhor aluno da turma.

```python
# 1. classe aluno
class Aluno:
    def __init__(self, nome, idade, curso):
        self.nome = nome
        self.idade = idade
        self.curso = curso
        self.notas = []

    def adicionar_nota(self, nota):
        self.notas += [nota]

    def media(self):
        if not self.notas:
            return 0
        return sum(self.notas) / len(self.notas)

    def exibir_info(self):
        print(f"Nome: {self.nome}, Idade: {self.idade}, Curso: {self.curso}, Média: {self.media():.2f}")


# 2. classe Turma
class Turma:
    def __init__(self, nome_turma):
        self.nome_turma = nome_turma
        self.alunos = []

    def adicionar_aluno(self, aluno):
        self.alunos += [aluno]

    def listar_alunos(self):
        print(f"\nAlunos da Turma '{self.nome_turma}':")
        for aluno in self.alunos:
            aluno.exibir_info()

    def melhor_aluno(self):
        if not self.alunos:
            return None

        melhor = self.alunos[0]
        for aluno in self.alunos[1:]:
            if aluno.media() > melhor.media():
                melhor = aluno
        return melhor


# 3. Fluxo do Programa

# criar pelo menos 3 alunos
aluno1 = Aluno("João", 20, "Engenharia")
aluno2 = Aluno("Diogo", 22, "Sistemas de informacao")
aluno3 = Aluno("Pedro", 21, "Direito")

# adicionar notas a cada aluno
aluno1.adicionar_nota(7.5)
aluno1.adicionar_nota(8.0)
aluno1.adicionar_nota(6.5)

aluno2.adicionar_nota(9.0)
aluno2.adicionar_nota(9.32)
aluno2.adicionar_nota(9.5)

aluno3.adicionar_nota(6.0)
aluno3.adicionar_nota(7.0)
aluno3.adicionar_nota(7.5)

# inserir os alunos em uma turma
turma1 = Turma("Turma A")
turma1.adicionar_aluno(aluno1)
turma1.adicionar_aluno(aluno2)
turma1.adicionar_aluno(aluno3)

# exibir os alunos da turma com suas medias
turma1.listar_alunos()

# mostrar o melhor aluno da turma
melhor_aluno = turma1.melhor_aluno()
if melhor_aluno:
    print(f"\nMelhor aluno da Turma '{turma1.nome_turma}':")
    melhor_aluno.exibir_info()
else:
    print(f"\nNão há alunos na Turma '{turma1.nome_turma}'.")
```
